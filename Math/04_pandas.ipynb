```python
df.drop_duplicates(inplace=True)
```

```python
df.dropna(subset = ['Description','CustomerID'], inplace =True)

df = df[(df.Quantity>0)]
```

```python
df.shape

# 5 lakh
# only 4 lakh 
```

```python
df["sales"] = df["Quantity"] * df["UnitPrice"]
df['Month'] = df['InvoiceDate'].dt.month
```

```python
# df.head(40)
def month_mapping(month):
    month_dict = {
        1: "January",
        2: "February",
        3: "March",
        4: "April",
        5: "May",
        6: "June",
        7: "July",
        8: "August",
        9: "September",
        10: "October",
        11: "November",
        12: "December"
    }
    return month_dict.get(month, "Invalid month")


df['Month'] = df['Month'].apply(month_mapping)
```

```python
# df.Month.value_counts()
top_10_product = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).reset_index().head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(top_10_product['Quantity'],bins=10)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_10_product['Description'],y=top_10_product['Quantity'])
plt.xticks(rotation=45)
plt.show()
```

```python
top_10_product.reset_index()
```

```python
# sales distribution ? 
product_sales = df.groupby('Description')['Quantity'].sum().sort_values(ascending = False).reset_index()

sns.histplot(product_sales['Quantity'],bins = 100, kde = True, edgecolor = '#FFFFFF')
plt.title('Product Distribution', fontsize = 16)
plt.xlabel('Quantity',fontsize = 12)
plt.ylabel('Count',fontsize = 12)
plt.show()
```

```python
import numpy as np
product_sales['log_quantity'] = np.log1p(product_sales['Quantity'])
sns.histplot(product_sales['log_quantity'], kde = True, edgecolor = '#FFFFFF')
plt.title('Log-Transformed Product Distribution')
plt.xlabel('Log(Quantity)')
plt.show()
```
