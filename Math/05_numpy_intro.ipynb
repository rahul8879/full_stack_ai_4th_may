```python
import random,time
```

```python
import numpy as np
N_DOCS= 500000
DIM = 384
docs = []

for i in range(N_DOCS):
    vectors = []
    for j in range(DIM):
        vectors.append(random.random())
    docs.append(vectors)

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

```

```python
# len(docs[0])

query= [random.random() for i in range(DIM)]
start = time.time()
for i in range(N_DOCS):
    cosine_similarity(query, docs[i])
print(time.time() - start)
```

```python
# memory computations
# in case of simplke list 100 documents * 384 dimensions * 8 bytes = 1.6 MB

docs_np= np.array(docs,dtype=np.float32)
query_np= np.array(query,dtype=np.float32)
results = []
start = time.time()
for i in range(N_DOCS):
    result = cosine_similarity(query_np, docs_np[i])
    results.append(result)
print(time.time() - start)
```

### Why so fast?
1. **Python loops are interpreted** — every `x*y` does type-checking + object overhead
2. **NumPy runs compiled C loops** over raw memory
3. **Contiguous memory + SIMD** — CPU crunches multiple floats per instruction
4. **`@` calls BLAS** — decades-optimized linear algebra libraries
