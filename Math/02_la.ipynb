```python
# remove rupees symbol from income
def clean_income(x):
    output = float(x.replace('₹','').replace(',',''))
    return output

data['income'] = data['income'].apply(clean_income)
data.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='cibil_score',y='income',data=data,hue='status')
plt.title('Cibil Score vs Income')
# plt.show()
plt.savefig('cibil_score_vs_income.png')
```

```python
# we need to find the line
w1 = -0.11685643
w2 = -0.00063651
w0 = 99

# find the line ( y=mx+c)
# w1*cibil_score + w2*income + w0=0

# income = -(w1*cibil_score + w0)/w2
income = -(w1*data['cibil_score'] + w0)/w2

# lets plot this line in scatter plot
# lets 

plt.figure(figsize=(10,6))
sns.scatterplot(x='cibil_score',y='income',data=data,hue='status')
sns.lineplot(x=data['cibil_score'],y=income,color='red')
```
