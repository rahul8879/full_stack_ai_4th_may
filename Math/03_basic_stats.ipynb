```python
import numpy as np


data = np.random.randint(0, 100, 10000)
print("Mean:", np.mean(data))  
```

```python
sample_1 = np.random.choice(data, 1000, replace=True)
sample_2 = np.random.choice(data, 1000, replace=True)
sample_3 = np.random.choice(data, 1000, replace=True)
```

```python
print("Sample 1 Mean:", np.mean(sample_1))
print("Sample 2 Mean:", np.mean(sample_2))
print("Sample 3 Mean:", np.mean(sample_3))

overall_mean = (   np.mean(sample_1) + np.mean(sample_2) + np.mean(sample_3) ) / 3
print("Overall Mean:", overall_mean)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(data, bins=30, kde=True)
plt.axvline(np.mean(data), color='r', linestyle='dashed', linewidth=1, label='Mean')
plt.legend()
plt.show()
```

```python
income = np.random.normal(50000, 15000, 1000)
print(np.mean(income))
print(np.std(income))
print(np.var(income))

min_1_std = np.mean(income) - 1 * np.std(income)
max_1_std = np.mean(income) + 1 * np.std(income)

min_2_std = np.mean(income) - 2 * np.std(income)
max_2_std = np.mean(income) + 2 * np.std(income)

min_3_std = np.mean(income) - 3 * np.std(income)
max_3_std = np.mean(income) + 3 * np.std(income)

# plot it
sns.histplot(income, bins=50, kde=True)

plt.axvline(np.mean(income), color='r', linestyle='dashed', linewidth=1, label='Mean')
plt.axvline(min_1_std, color='m', linestyle='dashed', linewidth=1, label='Min - 1*Std')
plt.axvline(max_1_std, color='c', linestyle='dashed', linewidth=1, label='Max + 1*Std')
plt.axvline(min_2_std, color='g', linestyle='dashed', linewidth=1, label='Min - 2*Std')
plt.axvline(max_2_std, color='b', linestyle='dashed', linewidth=1, label='Max + 2*Std')
plt.axvline(min_3_std, color='y', linestyle='dashed', linewidth=1, label='Min - 3*Std')
plt.axvline(max_3_std, color='c', linestyle='dashed', linewidth=1, label='Max + 3*Std')
plt.legend()
plt.show()
```

```python
import seaborn as sns
sns.set_theme(style="whitegrid")

penguins = sns.load_dataset("penguins")

# Draw a nested barplot by species and sex
g = sns.catplot(
    data=penguins, kind="bar",
    x="species", y="body_mass_g", hue="sex",
    errorbar="sd", palette="dark", alpha=.6, height=6
)
g.despine(left=True)
g.set_axis_labels("", "Body mass (g)")
g.legend.set_title("")
```
