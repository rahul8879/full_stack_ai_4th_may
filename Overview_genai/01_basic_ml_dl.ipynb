```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('customer_data.csv')
data.head(2)
```

```python
def clean_income(x):
    output = float(x.replace('₹','').replace(',',''))
    return output

data['Monthly Income (₹)'] = data['Monthly Income (₹)'].apply(clean_income)
data.head()
```

```python
def predict(df, new_customer):
    distances = []
    for index, row in df.iterrows():
        distance = np.sqrt((row['CIBIL Score'] - new_customer[0])**2 + (row['Monthly Income (₹)'] - new_customer[0])**2)
        distances.append((index, distance))

    # sort it 
    

    return distances
```

```python
new_customer = np.array([600,28])

distances = predict(data, new_customer)
distances
```
